### About the Dataset

**Note:** This project uses a synthetic dataset titled Indian Government Scheme Fund Utilization Data (2021–2025), sourced from Kaggle.

The dataset contains 1,000 artificially generated records representing hypothetical government scheme funding, fund utilization, beneficiary targets, and implementation status.

Although some scheme names and classifications are based on real government schemes, the financial figures, beneficiary counts, and individual records are artificially generated. Therefore, the findings in this analysis do not represent actual government spending or scheme performance.

**Purpose:** This dataset is used solely for educational purposes to practice data cleaning, exploratory data analysis (EDA) using Python , pandas and numpy.

```python
df['Date_of_Fund_Release'] = pd.to_datetime(df['Date_of_Fund_Release'],format="%Y-%m-%d")
df['Date_of_Fund_Release'].dtype
```

```python
df.groupby('State')['Fund_Released_Lakhs'].sum().sort_values(ascending=False).head(5)
```

**business insight:** West Bengal received the highest fund release (35,791 Lakhs), followed by Tamil Nadu (34,074 Lakhs). The gap between the two is only about 1,717 Lakhs (around 5%). The top five states (West Bengal, Tamil Nadu, Kerala, Bihar and Rajasthan) all fall between 30,000 and 36,000 Lakhs, so no single state is receiving a disproportionately large share of the funds

```python
amount_by_department = df.groupby('Department')[['Sanctioned_Amount_Lakhs','Fund_Released_Lakhs','Fund_Utilized_Lakhs']].sum().sort_values(by='Sanctioned_Amount_Lakhs',ascending=False)
amount_by_department
```

```python
df['released_Ratio'] = df['Fund_Released_Lakhs']/df['Sanctioned_Amount_Lakhs']
df.groupby('Department')['released_Ratio'].mean().sort_values(ascending=False)
```

```python
df['Utilization_Ratio'] = df['Fund_Utilized_Lakhs']/df['Fund_Released_Lakhs']
df.groupby('Department')['Utilization_Ratio'].mean().sort_values(ascending=False)
```

**business insight:** as we can see that the released ratio is between 74-76% but the Utilization Ratio is 61-67% . Rural Development received the most funds, yet roughly 28,090 Lakhs of its released amount remains unspent. Education has the lowest utilization (61.2%) and Health & Family Welfare the highest (67.5%). This suggests the bigger problem is not releasing funds but spending them, possibly because of delays in project execution. This needs to be checked against the Status and Audit_Remarks columns.

```python
df.groupby('Status')['Utilization_Percent'].mean()
```

***business insight:*** Of the 1,000 records, 349 are Completed, 330 Ongoing, 213 Delayed and 108 Not Started. Completed projects have an average utilization of about 88.3%, while Ongoing projects are at 63.0% and Delayed projects at 59.2%. Not Started projects have 0% utilization, as expected. About a third of all projects (321 records) are either Delayed or Not Started, so a large share of funds is not being put to work. Delayed projects utilize only about 4 percentage points less than Ongoing ones, so delay alone may not explain the low utilization seen in Q7. The Audit_Remarks column should be checked for other reasons.
